```python
#!/usr/bin/env python3
from dataclasses import dataclass, asdict
from datetime import datetime
import os, json, math, sys

try:
    from dateutil.relativedelta import relativedelta
    HAS_DATEUTIL = True
except ImportError:
    HAS_DATEUTIL = False

@dataclass
class Room:
    number: str
    room_type: str
    price_per_night: float
    features: str
    is_allocated: bool = False
    customer_name: str = None
    nights: int = 0
    check_in: str = None
    notes: str = None

    def to_dict(self):
        return asdict(self)

    @staticmethod
    def from_dict(d: dict):
        return Room(**d)

class LanghamHMS:
    def __init__(self, student_id: str):
        self.sid = (student_id or "00000000").strip()
        self.rooms = []
        self.file = f"LHMS_{self.sid}.txt"

    def find_room(self, num: str):
        for i, r in enumerate(self.rooms):
            if r.number == num:
                return i
        return -1

    def get_input(self, msg: str):
        try:
            return input(msg)
        except EOFError:
            print("\n[!] Input stream ended. Exiting.")
            self.exit_app()
            raise

    def to_float(self, s: str, default=None):
        try:
            return float(s)
        except ValueError:
            if default is not None:
                return default
            raise

    def to_int(self, s: str, default=None):
        try:
            return int(s)
        except ValueError:
            if default is not None:
                return default
            raise

    def add_room(self):
        try:
            num = self.get_input("Enter room number (e.g., 101): ").strip()
            if not num:
                raise ValueError("Room number cannot be empty.")

            if self.find_room(num) != -1:
                print("Room already exists.")
                return

            rtype = self.get_input("Enter room type (Single/Double/Suite): ").strip() or "Single"
            price = self.to_float(self.get_input("Enter price per night: ").strip())
            feat = self.get_input("Enter features (comma-separated): ").strip() or "WiFi, TV"

            self.rooms.append(Room(number=num, room_type=rtype, price_per_night=price, features=feat))
            print(f"[+] Room {num} added.")
        except ValueError as e:
            print(f"[ValueError] {e}")
        except TypeError as e:
            print(f"[TypeError] {e}")

    def delete_room(self):
        num = self.get_input("Enter room number to delete: ").strip()
        idx = self.find_room(num)
        if idx == -1:
            print("[!] Room not found.")
            return
        if self.rooms[idx].is_allocated:
            print("[!] Cannot delete: room is currently allocated.")
            return
        del self.rooms[idx]
        print(f"[-] Room {num} deleted.")

    def display_rooms(self):
        if not self.rooms:
            print("[i] No rooms available.")
            return
        print("\n--- Room Details ---")
        for r in self.rooms:
            status = "Allocated" if r.is_allocated else "Available"
            print(f"#{r.number} | {r.room_type:<6} | ${r.price_per_night:.2f}/night | {status} | Features: {r.features}")

    def allocate_room(self):
        num = self.get_input("Enter room number to allocate: ").strip()
        idx = self.find_room(num)
        if idx == -1:
            print("[!] Room not found.")
            return
        rm = self.rooms[idx]
        if rm.is_allocated:
            print("[!] Room already allocated.")
            return
        try:
            cust = self.get_input("Enter customer name: ").strip()
            nts = self.to_int(self.get_input("Enter number of nights: ").strip())
            if nts <= 0:
                raise ValueError("Nights must be positive.")
            chk = datetime.now().isoformat(timespec="seconds")
            rm.is_allocated = True
            rm.customer_name = cust
            rm.nights = nts
            rm.check_in = chk
            print(f"[+] Room {num} allocated to {cust} for {nts} night(s).")
        except ValueError as e:
            print(f"[ValueError] {e}")

    def display_allocations(self):
        print("\n--- Current Allocations ---")
        found = False
        for r in self.rooms:
            if r.is_allocated:
                found = True
                print(f"Room {r.number} -> {r.customer_name}, Nights: {r.nights}, Check-in: {r.check_in}")
        if not found:
            print("[i] No current allocations.")

    def calc_discount(self, expr: str, sub: float):
        allowed = {"subtotal": sub}
        try:
            disc = eval(expr, {"__builtins__": {}}, allowed)
            if not isinstance(disc, (int, float)):
                raise TypeError("Discount expression must evaluate to a number.")
            return float(disc)
        except (SyntaxError, NameError, TypeError) as e:
            print(f"[Discount Expression Error] {e.__class__.__name__}: {e}")
            return 0.0

    def deallocate_room(self):
        num = self.get_input("Enter room number to de-allocate: ").strip()
        idx = self.find_room(num)
        if idx == -1:
            print("[!] Room not found.")
            return
        rm = self.rooms[idx]
        if not rm.is_allocated:
            print("[!] Room is not currently allocated.")
            return
        
        rm.is_allocated = False
        rm.customer_name = None
        rm.nights = 0
        rm.check_in = None
        print(f"[✓] Room {num} has been de-allocated successfully.")

    def billing(self):
        num = self.get_input("Enter room number for billing: ").strip()
        idx = self.find_room(num)
        if idx == -1:
            print("[!] Room not found.")
            return
        rm = self.rooms[idx]
        if not rm.is_allocated:
            print("[!] Room is not allocated.")
            return

        try:
            nts = rm.nights
            try:
                avg = (rm.price_per_night * nts) / (nts - nts)
            except ZeroDivisionError:
                avg = 0.0

            sub = rm.price_per_night * nts

            expr = self.get_input("Enter discount expression using 'subtotal' (e.g., 'subtotal*0.1') or leave blank: ").strip()
            disc = self.calc_discount(expr, sub) if expr else 0.0

            try:
                fee = min(math.exp(2) / 100.0, 10.0)
            except OverflowError:
                fee = 10.0

            total = max(sub - disc + fee, 0.0)

            print(f"--- BILL FOR ROOM {rm.number} ---")
            print(f"Customer   : {rm.customer_name}")
            print(f"Nights     : {nts}")
            print(f"Rate/night : ${rm.price_per_night:.2f}")
            print(f"Subtotal   : ${sub:.2f}")
            print(f"Discount   : -${disc:.2f}")
            print(f"ServiceFee : ${fee:.2f}")
            print(f"TOTAL DUE  : ${total:.2f}")
        except (TypeError, ValueError) as e:
            print(f"[Billing Error] {e}")

    def save_to_file(self):
        try:
            data = [r.to_dict() for r in self.rooms]
            with open(self.file, "w", encoding="utf-8") as f:
                json.dump({"saved_at": datetime.now().isoformat(), "rooms": data}, f, indent=2)
            print(f"[✓] Saved to {self.file}")
        except FileNotFoundError as e:
            print(f"[FileNotFoundError] {e}")
        except IOError as e:
            print(f"[IOError] {e}")

    def display_from_file(self):
        try:
            with open(self.file, "r", encoding="utf-8") as f:
                content = f.read()
            print("\n--- FILE CONTENT START ---")
            print(content)
            print("--- FILE CONTENT END ---\n")
        except FileNotFoundError:
            print(f"[!] File not found: {self.file}")
        except IOError as e:
            print(f"[IOError] {e}")

    def backup_and_clear_file(self):
        try:
            try:
                with open(self.file, "r", encoding="utf-8") as f:
                    data = f.read()
            except FileNotFoundError:
                print(f"[!] Nothing to backup. '{self.file}' does not exist.")
                return

            ts = datetime.now().strftime("%Y%m%d_%H%M%S")
            bak = f"LHMS_{self.sid}_Backup_{ts}.txt"
            with open(bak, "a", encoding="utf-8") as bf:
                bf.write(data + "\n")
            with open(self.file, "w", encoding="utf-8") as f:
                f.truncate(0)
            print(f"[✓] Backed up to {bak} and cleared {self.file}.")
        except IOError as e:
            print(f"[IOError] {e}")

    def exit_app(self):
        print("Exiting Langham HMS. Goodbye!")
        sys.exit(0)

    def show_menu(self):
        print("\n---------------- Langham Hotel Management System ----------------")
        print("1. Add Rooms")
        print("2. Display Rooms")
        print("3. Allocate Rooms")
        print("4. De-Allocate Rooms")
        print("5. Display Room Allocation Details")
        print("6. Billing")
        print("7. Save the Room Allocations To a File")
        print("8. Show the Room Allocations From a File")
        print("9. Exit")
        print("------------------------------------------------------------------")

    def run(self):
        while True:
            self.show_menu()
            try:
                choice = self.get_input("Enter your choice number here: ").strip()
                if choice == "1":
                    self.add_room()
                elif choice == "2":
                    self.display_rooms()
                elif choice == "3":
                    self.allocate_room()
                elif choice == "4":
                    self.deallocate_room()
                elif choice == "5":
                    self.display_allocations()
                elif choice == "6":
                    self.billing()
                elif choice == "7":
                    self.save_to_file()
                elif choice == "8":
                    self.display_from_file()
                elif choice == "9":
                    self.exit_app()
                else:
                    try:
                        lst = [1,2,3]
                        _ = lst[10]
                    except IndexError:
                        print("[!] Invalid option. Please select 1-9.")
            except EOFError:
                print("\n[!] EOF received. Exiting.")
                self.exit_app()
            except Exception as e:
                print(f"[Unhandled Error] {e.__class__.__name__}: {e}")

def main():
    print("Welcome to Langham HMS")
    try:
        sid = input("Enter your Student ID (for file naming): ").strip()
    except EOFError:
        sid = "00000000"
    app = LanghamHMS(student_id=sid)
    app.run()

if __name__ == "__main__":
    main()
```
