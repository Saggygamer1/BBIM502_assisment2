##### from dataclasses import dataclass, asdict
from datetime import datetime

@dataclass
class Room:
    number: str
    room_type: str
    price_per_night: float
    features: str
    is_allocated: bool = False
    customer_name: str = None
    nights: int = 0
    check_in: str = None

class LanghamHMS:
    def __init__(self):
        self.rooms = []

    def find_room(self, num: str):
        for i, r in enumerate(self.rooms):
            if r.number == num:
                return i
        return -1

    def add_room(self):
        num = input("Enter room number (e.g., 101): ").strip()
        if self.find_room(num) != -1:
            print("[!] Room already exists.")
            return
        rtype = input("Enter room type (Single/Double/Suite): ").strip() or "Single"
        price = float(input("Enter price per night: ").strip())
        feat = input("Enter features (comma-separated): ").strip() or "WiFi, TV"
        self.rooms.append(Room(num, rtype, price, feat))
        print(f"[+] Room {num} added successfully!")

    def delete_room(self):
        num = input("Enter room number to delete: ").strip()
        idx = self.find_room(num)
        if idx == -1:
            print("[!] Room not found.")
            return
        if self.rooms[idx].is_allocated:
            print("[!] Room currently allocated. Cannot delete.")
            return
        del self.rooms[idx]
        print(f"[-] Room {num} deleted successfully!")

    def display_rooms(self):
        if not self.rooms:
            print("[i] No rooms available.")
        else:
            print("\n--- ROOM DETAILS ---")
            for r in self.rooms:
                status = "Allocated" if r.is_allocated else "Available"
                print(f"{r.number} | {r.room_type} | ${r.price_per_night:.2f}/night | {status} | Features: {r.features}")

    def allocate_room(self):
        num = input("Enter room number to allocate: ").strip()
        idx = self.find_room(num)
        if idx == -1:
            print("[!] Room not found.")
            return
        r = self.rooms[idx]
        if r.is_allocated:
            print("[!] Room already allocated.")
            return
        cust = input("Enter customer name: ").strip()
        nts = int(input("Enter number of nights: ").strip())
        r.is_allocated = True
        r.customer_name = cust
        r.nights = nts
        r.check_in = datetime.now().isoformat(timespec="seconds")
        print(f"[+] Room {num} allocated to {cust} for {nts} nights.")

    def display_allocations(self):
        print("\n--- CURRENT ALLOCATIONS ---")
        allocated = [r for r in self.rooms if r.is_allocated]
        if not allocated:
            print("[i] No allocations yet.")
        else:
            for r in allocated:
                print(f"Room {r.number} → {r.customer_name}, Nights: {r.nights}, Check-in: {r.check_in}")

    def billing_and_deallocate(self):
        num = input("Enter room number to bill & de-allocate: ").strip()
        idx = self.find_room(num)
        if idx == -1:
            print("[!] Room not found.")
            return
        r = self.rooms[idx]
        if not r.is_allocated:
            print("[!] Room is not allocated.")
            return
        discount = float(input("Enter discount (e.g., 0.1 for 10%): ") or 0)
        subtotal = r.price_per_night * r.nights
        total = subtotal - (subtotal * discount) + 10
        print(f"\n--- BILL FOR ROOM {r.number} ---")
        print(f"Customer   : {r.customer_name}")
        print(f"Nights     : {r.nights}")
        print(f"Rate/night : ${r.price_per_night}")
        print(f"Subtotal   : ${subtotal}")
        print(f"Discount   : -${subtotal * discount}")
        print(f"ServiceFee : $10.0")
        print(f"TOTAL DUE  : ${total}")
        r.is_allocated = False
        r.customer_name = None
        r.nights = 0
        print(f"[✓] Room {num} now de-allocated.")

    def run(self):
        while True:
            print("\n----- Langham Hotel Management System (Part A) -----")
            print("1. Add Room")
            print("2. Delete Room")
            print("3. Display Rooms")
            print("4. Allocate Room")
            print("5. Display Room Allocation Details")
            print("6. Billing & De-Allocation")
            print("0. Exit")
            print("----------------------------------------------------")

            choice = input("Enter option (0-6): ").strip()
            if choice == "1":
                self.add_room()
            elif choice == "2":
                self.delete_room()
            elif choice == "3":
                self.display_rooms()
            elif choice == "4":
                self.allocate_room()
            elif choice == "5":
                self.display_allocations()
            elif choice == "6":
                self.billing_and_deallocate()
            elif choice == "0":
                print("Exiting Langham HMS. Goodbye!")
                break
            else:
                print("[!] Invalid choice. Please select 0–6.")

app = LanghamHMS()
app.run()


```python
from dataclasses import dataclass, asdict
from datetime import datetime
import json, os

@dataclass
class Room:
    number: str
    room_type: str
    price_per_night: float
    features: str
    is_allocated: bool = False
    customer_name: str = None
    nights: int = 0
    check_in: str = None

    def to_dict(self):
        return asdict(self)

class LanghamFileIO:
    def __init__(self, student_id):
        self.sid = student_id
        self.file = f"LHMS_{self.sid}.txt"

    def save_to_file(self, rooms):
        data = [r.to_dict() for r in rooms]
        with open(self.file, "w", encoding="utf-8") as f:
            json.dump({"saved_at": datetime.now().isoformat(), "rooms": data}, f, indent=2)
        print(f"[✓] Data saved to {self.file}")

    def display_from_file(self):
        if not os.path.exists(self.file):
            print("[!] File not found.")
            return
        with open(self.file, "r", encoding="utf-8") as f:
            print("\n--- FILE CONTENT START ---")
            print(f.read())
            print("--- FILE CONTENT END ---")

    def backup_and_clear_file(self):
        if not os.path.exists(self.file):
            print("[!] Nothing to backup.")
            return
        ts = datetime.now().strftime("%Y%m%d_%H%M%S")
        bak = f"LHMS_{self.sid}_Backup_{ts}.txt"
        with open(self.file, "r", encoding="utf-8") as f:
            data = f.read()
        with open(bak, "a", encoding="utf-8") as bf:
            bf.write(data + "\n")
        open(self.file, "w").close()
        print(f"[✓] Backed up to {bak} and cleared {self.file}")

def main():
    sid = input("Enter your Student ID: ").strip()
    app = LanghamFileIO(sid)

    rooms = []
    while True:
        print("\n----- Langham Hotel File I/O System (Part B) -----")
        print("7. Save Room Allocations to File")
        print("8. Display Room Allocations from File")
        print("9. Backup and Clear File")
        print("0. Exit")
        print("--------------------------------------------------")

        choice = input("Enter option (0,7–9): ").strip()
        if choice == "7":
            # Example: enter 1-2 rooms quickly
            number = input("Enter room number: ")
            room_type = input("Enter room type: ")
            price = float(input("Enter price per night: "))
            features = input("Enter features: ")
            rooms.append(Room(number, room_type, price, features))
            app.save_to_file(rooms)
        elif choice == "8":
            app.display_from_file()
        elif choice == "9":
            app.backup_and_clear_file()
        elif choice == "0":
            print("Exiting File I/O system. Goodbye!")
            break
        else:
            print("[!] Invalid choice. Please select 0,7–9.")

main()
```

```python
from dataclasses import dataclass, asdict
from datetime import datetime
import json, os, math, sys

# ---- Optional Import Check (ImportError Handling) ----
try:
    from dateutil.relativedelta import relativedelta
except ImportError:
    print("[ImportError] Optional module 'dateutil' not found. Continuing without it.")

# ---- Data Class ----
@dataclass
class Room:
    number: int
    room_type: str
    price_per_night: float
    features: str
    is_allocated: bool = False
    customer_name: str = None
    nights: int = 0
    check_in: str = None

    def to_dict(self):
        return asdict(self)

# ---- Core HMS Class ----
class LanghamHMS:
    def __init__(self, sid):
        self.sid = sid
        self.rooms = []
        self.file = f"LHMS_{sid}.txt"

    # ---------------- Helper: Restart ----------------
    def restart(self):
        """Restart the menu after any handled error."""
        print("\n[↻] Restarting system menu...\n")
        run_menu(self)

    # ---------------- Add Room ----------------
    def add_room(self):
        try:
            # --- Room number validation ---
            while True:
                num_input = input("Enter room number (numbers only): ").strip()
                if num_input.isdigit():
                    num = int(num_input)
                    break
                else:
                    print("[!] Please enter a valid numeric room number.")

            if any(r.number == num for r in self.rooms):
                raise ValueError("Room already exists.")

            rtype = input("Enter room type (Single/Double/Suite): ") or "Single"

            # --- Price validation ---
            while True:
                price_input = input("Enter price per night: ").strip()
                if price_input.replace('.', '', 1).isdigit():
                    price = float(price_input)
                    break
                else:
                    print("[!] Please enter a valid number for price.")

            feat = input("Enter features: ") or "WiFi, TV"
            self.rooms.append(Room(num, rtype, price, feat))
            print(f"[+] Room {num} added successfully.")

        except ValueError as e:
            print(f"[ValueError] {e}")
            self.restart()
        except EOFError:
            print("[EOFError] Input stream closed unexpectedly.")
            self.restart()

    # ---------------- Allocate Room ----------------
    def allocate_room(self):
        try:
            while True:
                num_input = input("Enter room number to allocate: ").strip()
                if num_input.isdigit():
                    num = int(num_input)
                    break
                else:
                    print("[!] Room number must be numeric.")

            r = next((x for x in self.rooms if x.number == num), None)
            if not r:
                raise IndexError("Room not found.")
            if r.is_allocated:
                raise TypeError("Room already allocated.")
            cust = input("Enter customer name: ")

            # --- Nights validation ---
            while True:
                nights_input = input("Enter number of nights: ").strip()
                if nights_input.isdigit():
                    nights = int(nights_input)
                    if nights > 0:
                        break
                    else:
                        print("[!] Nights must be greater than zero.")
                else:
                    print("[!] Please enter a valid integer for nights.")

            r.is_allocated = True
            r.customer_name = cust
            r.nights = nights
            r.check_in = datetime.now().isoformat(timespec="seconds")
            print(f"[+] Room {num} allocated to {cust}.")
        except (ValueError, TypeError, IndexError) as e:
            print(f"[Handled Error] {e}")
            self.restart()

    # ---------------- Billing & Deallocate ----------------
    def billing_and_deallocate(self):
        try:
            while True:
                num_input = input("Enter room number for billing: ").strip()
                if num_input.isdigit():
                    num = int(num_input)
                    break
                else:
                    print("[!] Room number must be numeric.")

            r = next((x for x in self.rooms if x.number == num), None)
            if not r:
                raise NameError("Room not found.")
            if not r.is_allocated:
                raise TypeError("Room is not allocated.")

            # --- Discount validation ---
            while True:
                disc_input = input("Enter discount (0.1 = 10%): ").strip() or "0"
                try:
                    disc = float(disc_input)
                    if 0 <= disc < 1:
                        break
                    else:
                        print("[!] Enter a valid decimal between 0 and 1.")
                except ValueError:
                    print("[!] Please enter a valid numeric discount value.")

            subtotal = r.price_per_night * r.nights
            try:
                avg = subtotal / (r.nights - r.nights)  # triggers ZeroDivisionError
            except ZeroDivisionError:
                avg = 0
            try:
                fee = min(math.exp(2)/100, 10)
            except OverflowError:
                fee = 10

            total = subtotal - (subtotal * disc) + fee
            print(f"\n--- BILL ---")
            print(f"Subtotal: ${subtotal:.2f}  Discount: ${subtotal*disc:.2f}  Fee: ${fee:.2f}")
            print(f"Total Due: ${total:.2f}")
            r.is_allocated = False
            r.customer_name = None
            r.nights = 0
            print(f"[✓] Room {num} de-allocated.")
        except (NameError, TypeError, OverflowError) as e:
            print(f"[Billing Error] {e}")
            self.restart()

    # ---------------- File Operations ----------------
    def save_to_file(self):
        try:
            data = [r.to_dict() for r in self.rooms]
            with open(self.file, "w", encoding="utf-8") as f:
                json.dump(data, f, indent=2)
            print(f"[✓] Saved to {self.file}")
        except (IOError, FileNotFoundError) as e:
            print(f"[File Error] {e}")
            self.restart()

    def display_from_file(self):
        try:
            with open(self.file, "r", encoding="utf-8") as f:
                print("\n--- FILE CONTENT ---")
                print(f.read())
        except FileNotFoundError:
            print("[FileNotFoundError] File does not exist.")
            self.restart()
        except IOError:
            print("[IOError] Unable to open file.")
            self.restart()

    def backup_and_clear_file(self):
        try:
            if not os.path.exists(self.file):
                raise FileNotFoundError("Main file missing.")
            ts = datetime.now().strftime("%Y%m%d_%H%M%S")
            bak = f"LHMS_{self.sid}_Backup_{ts}.txt"
            with open(self.file, "r", encoding="utf-8") as f:
                data = f.read()
            with open(bak, "a", encoding="utf-8") as bf:
                bf.write(data + "\n")
            open(self.file, "w").close()
            print(f"[✓] Backed up to {bak} and cleared main file.")
        except (FileNotFoundError, IOError) as e:
            print(f"[Backup Error] {e}")
            self.restart()

# ---------------- Menu Function ----------------
def run_menu(app):
    print("\n--- MENU (choose one per run) ---")
    print("1.Add  2.Allocate  3.Billing  4.Save  5.DisplayFile  6.Backup  0.Exit")

    try:
        while True:
            choice = input("Enter option (0-6): ").strip()
            if choice.isdigit() and 0 <= int(choice) <= 6:
                break
            else:
                print("[!] Please enter a number between 0 and 6.")

        if choice == "1": app.add_room()
        elif choice == "2": app.allocate_room()
        elif choice == "3": app.billing_and_deallocate()
        elif choice == "4": app.save_to_file()
        elif choice == "5": app.display_from_file()
        elif choice == "6": app.backup_and_clear_file()
        elif choice == "0":
            print("Exiting system. Goodbye!")
    except EOFError:
        print("[EOFError] Input stream ended.")
        app.restart()
    except SyntaxError:
        print("[SyntaxError] Invalid code structure.")
        app.restart()

# ---------------- Start Program ----------------
while True:
    sid_input = input("Enter Student ID (numbers only): ").strip()
    if sid_input.isdigit():
        sid = sid_input
        break
    else:
        print("[!] Student ID must be numeric digits only.")

app = LanghamHMS(sid)
run_menu(app)
```
